```python
# Calculate the daily average percent future growth for base_data
base_data_avg_growth = base_data.groupby(level='Date')['Percent 1 Week Growth'].mean()

# Calculate the daily average percent future growth for sub_data
gainers_avg_growth = gainers_data.groupby(level='Date')['Percent 1 Week Growth'].mean()


# Compare the average percent future growth between base_data and sub_data
comparison = base_data_avg_growth < gainers_avg_growth
comparison.sum()
```

```python
losers_avg_growth = losers_data.groupby(level='Date')['Percent 1 Week Growth'].mean()


# Compare the average percent future growth between base_data and sub_data
comparison = base_data_avg_growth < losers_avg_growth
comparison.sum()
```
