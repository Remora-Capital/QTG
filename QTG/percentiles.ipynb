```python
import pandas as pd
import os

chosen_year = "2023"


data = pd.read_pickle(f'./snp/{chosen_year}.pkl')
data
```

```python
data["Day Percent Change"] = (data["Close"] - data["Open"]) / data["Open"]
```

```python
# Calculate the change between each day's close and the previous day's close for each ticker
data['Previous Close'] = data.groupby(level='Ticker')['Close'].shift(1)
data['Percent Change'] = (data['Close'] - data['Previous Close']) / data['Previous Close']
data
```

```python
data["Next Day Close"] = data.groupby(level='Ticker')['Close'].shift(-1)
data['Next Day Percent Change'] = (data['Next Day Close'] - data['Close']) / data['Close']
data
```

```python
# Calculate the change between each day's close and the previous day's close for each ticker
data['2 Week Close'] = data.groupby(level='Ticker')['Close'].shift(-10)
data['1 Week Close'] = data.groupby(level="Ticker")['Close'].shift(-5)
data['Percent 2 Week Growth'] = (data['2 Week Close'] - data['Close']) / data['Close']
data['Percent 1 Week Growth'] = (data['1 Week Close'] - data['Close']) / data['Close']
data["Percent 1 to 2 Week Growth"] = (data['2 Week Close'] - data['1 Week Close']) / data['1 Week Close']
data
```

```python
data["6 Days Prev Close"] = data.groupby(level='Ticker')['Close'].shift(6)
data["6 Days Prev Change"] = (data["Previous Close"] - data["6 Days Prev Close"]) / data["6 Days Prev Close"]
data["1 Month Prev Close"] = data.groupby(level='Ticker')['Close'].shift(24)
data["1 Month Prev Change"] = (data["Previous Close"] - data["1 Month Prev Close"]) / data["1 Month Prev Close"]
data["3 Months Prev Close"] = data.groupby(level='Ticker')['Close'].shift(72)
data["3 Months Prev Change"] = (data["Previous Close"] - data["3 Months Prev Close"]) / data["3 Months Prev Close"]
data["6 Months Prev Close"] = data.groupby(level='Ticker')['Close'].shift(148)
data["6 Months Prev Change"] = (data["Previous Close"] - data["6 Months Prev Close"]) / data["6 Months Prev Close"]
```

```python
data
```

```python
sorted_data = data.groupby(level=0, group_keys=False).apply(lambda x: x.sort_values('Percent Change', ascending=False))
sorted_data
```

```python
# Get the top 25 stocks by percent change for each day
top_25_stocks = sorted_data.groupby(level=0, group_keys=False).apply(lambda x: x.nlargest(25, 'Percent Change'))

# Get the bottom 25 stocks by percent change for each day
bottom_25_stocks = sorted_data.groupby(level=0, group_keys=False).apply(lambda x: x.nsmallest(25, 'Percent Change'))

top_25_stocks
```

```python
data['Gainer'] = data.index.isin(top_25_stocks.index)
data['Loser'] = data.index.isin(bottom_25_stocks.index)
data.to_pickle(f'./labelled_market/{chosen_year}.pkl')
```

```python
top_25_stocks_percentiles_self = [top_25_stocks.groupby(level=0, group_keys=False).apply(lambda x: x.iloc[i-5:i]) for i in range(5, 30, 5)]
bottom_25_stocks_percentiles_self = [bottom_25_stocks.groupby(level=0, group_keys=False).apply(lambda x: x.iloc[i-5:i]) for i in range(5, 30, 5)]

for i, percentile in enumerate(top_25_stocks_percentiles_self):
    percentile.to_pickle(f"./gain_lose_percentiles/{chosen_year}_gainers_{(i+1)*20}.pkl")
    print(percentile.mean()["Percent Change"])

for i, percentile in enumerate(bottom_25_stocks_percentiles_self):
    percentile.to_pickle(f"./gain_lose_percentiles/{chosen_year}_losers_{(i+1)*20}.pkl")
    print(percentile.mean()["Percent Change"])
```
